# Phase 2 — Data Exploration & Verification

This notebook explores the processed hybrid simulation dataset:
- IEEE 33-bus load profiles (`data/processed/load_profiles.parquet`)
- Synthetic anomaly labels (`data/processed/anomaly_labels.parquet`)
- Temporal train/validation/test splits (`data/processed/splits.json`)


```python
from pathlib import Path
import pandas as pd
from src.utils.io import load_parquet, load_json

# Load processed dataset and manifest
df_profiles = load_parquet("data/processed/load_profiles.parquet")
df_labels = load_parquet("data/processed/anomaly_labels.parquet")
splits = load_json("data/processed/splits.json")
manifest = load_json("data/processed/manifest.json")

print(f"Total Timesteps: {len(df_profiles)}")
print(f"Total Features: {len(df_profiles.columns)}")
print(f"Anomalous Steps: {df_labels['is_anomaly'].sum()} ({df_labels['is_anomaly'].mean()*100:.2f}%)")
print(f"Dataset Designation: {manifest.get('dataset_designation')}")
```

```python
# Display summary statistics of feeder active load across all 32 buses
bus_p_cols = [c for c in df_profiles.columns if c.startswith("bus_") and c.endswith("_p_kw")]
total_feeder_p = df_profiles[bus_p_cols].sum(axis=1)
print("Feeder Total Active Load (kW) Summary:")
print(total_feeder_p.describe())
```

```python
# Inspect anomaly type breakdown
print("Anomaly Type Distribution:")
print(df_labels["anomaly_type"].value_counts())
```

```python
# Verify temporal split chronological ordering
for split_name, details in splits["date_ranges"].items():
    print(f"{split_name.capitalize():12}: {details['start']} to {details['end']} ({details['count']} steps)")
```
