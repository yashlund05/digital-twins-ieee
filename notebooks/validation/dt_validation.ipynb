# Experiment E1 — Digital Twin Baseline Validation

This notebook executes and analyzes the baseline validation for the OpenDSS IEEE 33-bus Digital Twin:
- AC power flow solution on standard Baran & Wu (1989) benchmark network
- Voltage profile analysis across all 33 buses (verifying [0.90, 1.10] pu bounds)
- Power balance conservation law verification (|P_gen - P_load - P_loss| < 0.5%)
- Multi-condition testing: Light (0.5x), Nominal (1.0x), and Heavy (1.5x) loading


```python
from pathlib import Path

import pandas as pd

from src.digital_twin.initializer import run_experiment_e1_validation

# Run Experiment E1
result = run_experiment_e1_validation(seed=42)
print(f"Status: {result['status']}")
print(f"Run ID: {result['run_id']}")
print(f"All Validation Criteria Passed: {result['all_passed']}")
```

```python
# Load and display power flow summary across conditions
run_dir = Path(result["run_dir"])
df_power = pd.read_csv(run_dir / "results" / "power_flow.csv")
print("Power Flow Across Loading Conditions:")
print(
    df_power[
        [
            "condition",
            "multiplier",
            "total_gen_p_kw",
            "total_load_p_kw",
            "total_loss_p_kw",
            "min_v_pu",
            "balance_err_pct",
        ]
    ]
)
```

```python
# Inspect per-bus voltage profile for nominal condition
df_voltages = pd.read_csv(run_dir / "results" / "voltage_profile.csv")
df_nom_v = df_voltages[df_voltages["condition"] == "nominal_1.0x"].sort_values("bus_id")
print("Nominal Voltage Profile (Selected Buses):")
print(
    df_nom_v[df_nom_v["bus_id"].isin([1, 2, 6, 12, 18, 22, 25, 33])][
        ["bus_id", "v_pu", "v_kv", "angle_deg"]
    ]
)
```
